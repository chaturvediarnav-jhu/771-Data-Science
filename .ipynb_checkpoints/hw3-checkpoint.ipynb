```python
import pandas as pd 
import numpy as np 
import matplotlib as mpl
import seaborn as sns
import matplotlib.pyplot as plt
sns.set()

# bring in the data set
df=pd.read_csv("https://raw.githubusercontent.com/smart-stats/ds4bio_book/main/book/assetts/kirby21AllLevels.csv")

# isolate subject kirby90a_ax.img
kirby906a_ax_df = df.loc[(df['rawid']=='kirby906a_ax.img')].copy()

# do calculations for type 1 level 1 Telencephalon L and R volume, and ICV and TBV fractions
telencephalon_L_volume = sum(kirby906a_ax_df['volume'][(kirby906a_ax_df['type']==1) & (kirby906a_ax_df['level']==1) & (kirby906a_ax_df['roi']=='Telencephalon_L')])
telencephalon_R_volume = sum(kirby906a_ax_df['volume'][(kirby906a_ax_df['type']==1) & (kirby906a_ax_df['level']==1) & (kirby906a_ax_df['roi']=='Telencephalon_R')])

kirby906a_ax_df['icv_fraction'] = kirby906a_ax_df['volume']/kirby906a_ax_df['icv']
ICV_fraction_L = sum(kirby906a_ax_df['icv_fraction'][(kirby906a_ax_df['type']==1) & (kirby906a_ax_df['level']==1) & (kirby906a_ax_df['roi']=='Telencephalon_L')])
ICV_fraction_R = sum(kirby906a_ax_df['icv_fraction'][(kirby906a_ax_df['type']==1) & (kirby906a_ax_df['level']==1) & (kirby906a_ax_df['roi']=='Telencephalon_R')])
ICV_fraction = ICV_fraction_L + ICV_fraction_R

kirby906a_ax_df['tbv_fraction'] = kirby906a_ax_df['volume']/kirby906a_ax_df['tbv']
TBV_fraction_L = sum(kirby906a_ax_df['tbv_fraction'][(kirby906a_ax_df['type']==1) & (kirby906a_ax_df['level']==1) & (kirby906a_ax_df['roi']=='Telencephalon_L')])
TBV_fraction_R = sum(kirby906a_ax_df['tbv_fraction'][(kirby906a_ax_df['type']==1) & (kirby906a_ax_df['level']==1) & (kirby906a_ax_df['roi']=='Telencephalon_R')])
TBV_fraction = TBV_fraction_L + TBV_fraction_R

# print results
print (telencephalon_L_volume)
print (telencephalon_R_volume)
print (ICV_fraction)
print (TBV_fraction)
```

```python
# data frame for all sum of all regions
all_regions_df = kirby906a_ax_df.groupby(['type','level'])['volume'].sum().reset_index()
print(all_regions_df)
```

```python
# barplot to show Type 1 Level 2 data
t1l2 = kirby906a_ax_df[(kirby906a_ax_df['type']==1) & (kirby906a_ax_df['level']==2)]

# make plot
t1l2plot = sns.barplot(x='roi', y='volume', data=t1l2)
plt.xticks(rotation=90)   # rotate so names don't overlap
plt.xlabel('Region (ROI)')
plt.ylabel('Volume')
plt.title("kirby906a_ax: Type 1, Level 2 Region Volumes")
plt.show
```

```python
#donut plot to show Type 1 Level 3 data
t1l3 = kirby906a_ax_df[(kirby906a_ax_df['type']==1) & (kirby906a_ax_df['level']==3)]
plt.figure(figsize=(20,20)) # enlarge to see data labels
t1l3plot, labels = plt.pie(t1l3['volume'], labels=t1l3['roi'])
# rotate so words overlap less
for t in labels:
    t.set_rotation(-20)    
plt.title("kirby906a_ax: Type 1, Level 3 Region Volumes")
center_circle = plt.Circle((0, 0), 0.70, fc='white')
plt.gca().add_artist(center_circle)
plt.show()
```

```python
# bring in the data set
url = ("https://raw.githubusercontent.com/bcaffo/ds4bme/master/data/classInterests.txt")
class_interests_df=pd.read_csv(url, sep='\s+') # add separation to split the data into two columns
plot = sns.countplot(x='Program',hue='Year', data=class_interests_df)
plt.xticks(rotation=90) # rotate text to improve readability
plt.xlabel('Program')
plt.ylabel('Number of Students')
plt.title('Number of Students in Program by Year')
plt.legend(title='Year')
plt.show()
```

```python
# bring in the data set
url = ("https://raw.githubusercontent.com/jhu-advdatasci/2018/master/data/GSE5859_exprs.csv")
gene_expression_df = pd.read_csv(url,index_col=0) # first column is row labels 

# subtract row mean from each row and name is gene-expression_df_pt1
rowmeans = gene_expression_df.mean(axis=1)
gene_expression_df_pt1 = gene_expression_df.sub(rowmeans, axis=0)
# simple check new row mean should = 0
print(gene_expression_df_pt1.mean(axis=1).max())

# take resulting matrix and subtract column means from each column and name it gene_expression_df_pt2
columnmeans = gene_expression_df_pt1.mean(axis=0)
gene_expression_df_pt2 = gene_expression_df_pt1.sub(columnmeans,axis=1)
# simple check new column mean should = 0
print(gene_expression_df_pt2.mean(axis=0).max())

# take the resulting matrix and divide each column by the columns std dev and name it gene_expression_df_pt3
std_devs = gene_expression_df_pt2.std(axis=0)
gene_expression_df_pt3 = gene_expression_df_pt2.div(std_devs,axis=1)
# simple check new row std dev should = 1
print(gene_expression_df_pt3.std(axis=0).max())
print(gene_expression_df_pt3.std(axis=0).min())
```

```python
# bring in the data set
url = ("https://raw.githubusercontent.com/jhu-advdatasci/2018/master/data/KFF/healthcare-spending.csv")
healthcare_df = pd.read_csv(url,skiprows=2) 

# only need the data rows so drop any rows with NaN values
healthcare_df = healthcare_df.dropna()

# only want healthcare by state so remove USA and DC rows
healthcare_df = healthcare_df[healthcare_df['Location'] != 'United States']
healthcare_df = healthcare_df[healthcare_df['Location'] != 'District of Columbia']


# Change column title names to just be the year 
healthcare_df.columns = [healthcare_df.columns[0]] + [col[:4] for col in healthcare_df.columns[1:]]

#make plot
longtable = healthcare_df.melt(id_vars = 'Location', var_name = 'Year', value_name='Spending')
plot = sns.lineplot(data=longtable,x='Year',y='Spending', hue='Location')
plt.xticks(rotation=90)
plt.xlabel('Year')
plt.ylabel('Healthcare Spending')
plt.title('Healthcare Spending over Time by State')
plt.legend(bbox_to_anchor=(1, 1), loc='upper left')  # legend outside, small font
plt.show()
```

```python
#make barplot of average healthcare spend by state 

# calculate the means while keeping location column
avgspend = healthcare_df.copy()
avgspend['AvgSpend'] = avgspend.drop(columns='Location').mean(axis=1)
avgspend = avgspend[['Location','AvgSpend']]

# make plot
plot = sns.barplot(x='Location', y='AvgSpend',data=avgspend)
plt.xticks(rotation=90)
plt.xlabel('State')
plt.xticks(size=8)
plt.ylabel('Healthcare Spending')
plt.title('Average Healthcare Spending by State from 1991 to 2014')
plt.show()
```
