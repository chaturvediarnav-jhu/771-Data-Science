```python
#1.Sorting Hat Program 
def sorting_hat ():
    x = input("Do you like the color green? (y/n) > ")
    y = input("Do you have blonde hair? y/n? >")
    if (x!= "y" and x!= "n") or (y!= "y" and y!= "n") :
        print ("Please enter y or n exactly")
    elif x == "y" and y == "y" :
        Out = ("Slytherine!")
    elif x == "n" and y == "n":
        Out = ("Griffindor!")
    elif x == "n" or y == "n":
        Out = ("Griffindor!")
    else :
        print("Please enter y or n exactly")
    return Out

sorting_hat()

    
```

```python
#2.mydata Problem
class mydata(set):
    def mean(self):
        return sum(self) / len(self)
        #return set.mean
    def std(self):
        dist_sq = []
        m = self.mean()
        for i in self: dist_sq.append((i-m)**2)
        return(sum(dist_sq)/len(self)) ** .5

#test case
x = mydata([1, 2, 7, 5])
print(x.mean())
print(x.std())
```

```python
#3.lag functions
import numpy
def lag(vec,num):
    vec = vec.astype(float)
    i = 0
    while i<num: 
        vec = numpy.concatenate([[numpy.nan], vec[:-1]])
        i = i+1
    return vec

#test case
a = numpy.array( [1, 6, 8, 4] )
print(a)
print(lag(a,1))
```

```python
#4.Lag generation problem
def lagGenerator(num):
    return lambda vec:lag(vec,num) 

lag_2 = lagGenerator(2)
print(lag_2(a))
lag_4 = lagGenerator(4)
print(lag_4(a))
```
