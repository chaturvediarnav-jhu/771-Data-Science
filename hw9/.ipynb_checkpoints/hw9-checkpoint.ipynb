```python
# Probelm 1: Covid Case Data
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn.linear_model as lm
import sklearn as skl
import statsmodels.formula.api as smf
import statsmodels as sm
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# bring in the data set
df=pd.read_csv("https://raw.githubusercontent.com/CSSEGISandData/COVID-19/master/csse_covid_19_data/csse_covid_19_time_series/time_series_covid19_confirmed_global.csv")

#isolate Brazil data
brazil = df[df['Country/Region'] == 'Brazil']
# print(brazil.shape)
cumul = brazil.iloc[0,4:].to_numpy(dtype=float)
daily = np.diff(cumul,prepend = 0) # find daily cases by taking difference of cases between days
fd = np.argmax(daily>0) #find first day
y = daily[fd:]

#build regression equation
T = len(y)
Y = y[5:]
X = np.column_stack([y[5-k:T-k] for k in range(1,6)])

# print(y[:20])# first term needs to be non zero
# print(Y.shape,X.shape)

dat = pd.DataFrame(X, columns=['B1','B2','B3','B4','B5'])
dat['Y']=Y

results = smf.ols('Y~B1+B2+B3+B4+B5', data = dat).fit()
print(results.summary2())
```

## Problem 1 Discussion

Intercept $= 3617.69$: This is the case count the model predicts for a day when the previous 5 days have 0 cases.

**$\beta_1 = 0.6067$**: When holding the rest of the coefficients, when one case is added on day $T-1$, $0.61$ cases are added on day $T$. This is a strong predictor given the low P value and high confidence interval.

**$\beta_2 = -0.0129$**: Confidence interval includes 0, therefore these predictors are not good.

**$\beta_3 = 0.0402$**: Confidence interval includes 0, therefore these predictors are not good.

**$\beta_4 = 0.0742$**: When holding the rest of the coefficients, when one case is added on day $T-4$, $0.07$ cases are added on day $T$. This has a low P value so there is some contribution from this coefficient but significantly less than $\beta_1$.

**$\beta_5 = 0.1842$**: When holding the rest of the coefficients, when one case is added on day $T-5$, $0.18$ cases are added on day $T$. This has a low P value so there is some contribution from this coefficient but significantly less than $\beta_1$.

```python
# Problem 2: Time vs pediction error
#build lagged data set
n=len(Y)
t=np.arange(5,5+n)

#fit on all of A, apply to all of A
fit_all = LinearRegression().fit(X,Y)
err_a = Y-fit_all.predict(X)
mse_a = np.mean(err_a**2)

#fit on first half, apply to second half
half = n//2
fit_half = LinearRegression().fit(X[:half],Y[:half])
err_b = Y[half:]-fit_half.predict(X[half:])
mse_b = np.mean(err_b**2)

fig, axes = plt.subplots(1, 2, figsize=(15, 4))

axes[0].plot(t, err_a)
axes[0].set_title(f'a. Fit on all, apply to all (MSE = {mse_a:.3g})')
axes[0].set_xlabel('Days since outbreak onset')
axes[0].set_ylabel('Prediction error (actual - predicted)')

axes[1].plot(t[half:], err_b)
axes[1].set_title(f'b. Fit on first half, apply to second half (MSE = {mse_b:.3g})')
axes[1].set_xlabel('Days since outbreak onset')
axes[1].set_ylabel('Prediction error (actual - predicted)')

plt.tight_layout()
plt.show()
```

```python
# Problem 3: Apply model for country A on country B
def model(country):
    row = df[df['Country/Region'] == country]
    cumul = row.iloc[0,4:].to_numpy(dtype=float)
    daily = np.diff(cumul,prepend = 0) # find daily cases by taking difference of cases between days
    fd = np.argmax(daily>0) #find first day
    y = daily[fd:]
    #build regression equation
    T = len(y)
    Y = y[5:]
    X = np.column_stack([y[5-k:T-k] for k in range(1,6)])
    return X, Y

#find parameters of country A and country B
XA,YA = model('Brazil')
XB,YB = model('Peru')

#fit on A
fitA = LinearRegression().fit(XA,YA)
#apply to B
errB = YB - fitA.predict(XB)
mseB = np.mean(errB**2)

t = np.arange(5,5+len(YB))

plt.figure(figsize=(8, 4))
plt.plot(t, errB)
plt.xlabel("Days since outbreak onset (Peru)")
plt.ylabel("Prediction error (actual - predicted)")
plt.title(f"Model fit on Brazil, applied to Peru (MSE = {mseB:.3g})")
plt.show()
```

```python
# Problem 4: logit model
#build Z 
z = (y[1:]>y[:-1]).astype(int)
Tz = len(z)
Zt = z[5:]
Z = np.column_stack([z[5 - k : Tz - k] for k in range(1, 6)])
dat = pd.DataFrame(Z, columns=['b1','b2','b3','b4','b5'])
dat['Z']=Zt
fit=smf.logit('Z~b1+b2+b3+b4+b5',data=dat).fit()
print(fit.summary())

odds = np.exp(fit.params)
table = pd.DataFrame([fit.params.rename('coef'), odds.rename('odds')])
print(table)
```

Intercept $= 1.3847$: The odds here are 4 to 1 meaning the likeliness that there will be an increase in cases after 5 days without an increase is 4 to 1.

**$\beta_1 = -0.2616$**: The odds here of an increase in cases after 1 day of no increase in cases is 0.77 to 1 but given the confidence interval includes 0, this coefficient may not have an effect.

**$\beta_2 = -0.8523$**: Increase in cases after 2 days of no increase is 0.43 to 1

**$\beta_3 = -0.9245$**: Increase in cases after 3 days of no increase is 0.39 to 1

**$\beta_4 = -0.7120$**: Increase in cases after 4 days of no increase is 0.49 to 1

**$\beta_5 = -0.7668$**: Increase in cases after 5 days of no increase is 0.46 to 1

Every coefficient is negative so a day where there is an increase in cases means it is unlikely that another day like that will happen.
